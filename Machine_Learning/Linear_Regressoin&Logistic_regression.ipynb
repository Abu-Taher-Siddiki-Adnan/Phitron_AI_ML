Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Create a sample dataset
hours = np.array([2,4,6,8,10]).reshape(-1,1)
marks = np.array([55,60,70,80,88])

hours, marks
```

Train a Linear Regression Model

```python
model = LinearRegression()
model.fit(hours, marks)
```

```python
print("m: ",model.coef_[0])
print("c: ",model.intercept_)
```

Visualize Best Fit line

```python
plt.scatter(hours, marks, color='blue', label ="Actual Data")
predicted_marks = model.predict(hours)
plt.plot(hours, predicted_marks, color='red', label = 'Best Fit Line')

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title('Study Hours vs Marks for Linear Regression')
plt.legend()
plt.grid()
plt.show()
```

Predict Marks for a new student

```python
new_hours = np.array([[7]])
prediction = model.predict(new_hours)
print('Predicted Marks', prediction[0])
```

##### Cost function and Gradinet Descent

```python
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import  mean_squared_error
```

```python
# Fit a Linear Regression model (Closed-Form Solver)

lin_reg = LinearRegression()
lin_reg.fit(hours, marks)

print("m: ",lin_reg.coef_[0])
print("c: ",lin_reg.intercept_)
```

```python
# Compute the cost (MSE) using a library
pre_marks = lin_reg.predict(hours)
mse_lin = mean_squared_error(marks, pre_marks)

print("MSE for linear regression: ", mse_lin)
```

```python
plt.scatter(hours, marks, color='blue', label ="Actual Data")

plt.plot(hours, pre_marks, color='red', label = 'Best Fit Line')

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title('Study Hours vs Marks for Linear Regression')
plt.legend()
plt.grid()
plt.show()
```

```python
# Use Gradient Descent Based Model - SGDRegressor
sgd_reg = SGDRegressor(max_iter=1000, learning_rate='invscaling', eta0=0.01, random_state=42)
sgd_reg.fit(hours, marks)

print('m: ', sgd_reg.coef_[0])
print('c: ', sgd_reg.intercept_[0])
```

```python
pred_marks_sgd = sgd_reg.predict(hours)
mse_sgd = mean_squared_error(marks, pred_marks_sgd)

print('MSE for SGD Regressor: ',mse_sgd)
```

```python
plt.scatter(hours, marks, color='blue', label ="Actual Data")

plt.plot(hours, pre_marks, color='green', label = 'Linear Regression Line', linestyle = '--')
plt.plot(hours, pred_marks_sgd, color='red', label = 'SGDRegressor Line', linestyle = '-.')

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title('Linear Regression vs SGDRegressor')
plt.legend()
plt.grid()
plt.show()
```

```python
# Evaluatoin Metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
new_model = LinearRegression()
new_model.fit(hours, marks)
new_pred = new_model.predict(hours)
print(new_pred)
```

```python
# MAE(Mean Absolute Error)
mae = mean_absolute_error(marks, new_pred)

# MSE(Mean Squared Error)
mse = mean_squared_error(marks, new_pred)

# RMSE(Root Mean Squared Error)
rmse = np.sqrt(mse)

# R2 
r2 = r2_score(marks, new_pred)
```

```python
print("MAE: ", mae)
print("MSE: ", mse)
print("RMSE: ", rmse)
print("R^2: ", r2)
```
