```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder,StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('loan_approval.csv')
df.head(10)
```

```python
df.shape
```

```python
print("Unique Values per column:")
df.nunique()
```

```python
print("Missing Value Per column")
df.isnull().sum()
```

Comment: There is no missing value in this dataset

```python
print("Column Data Type")
display(df.dtypes)
```

```python
le = LabelEncoder()
df['loan_approved'] = le.fit_transform(df['loan_approved'])
df.head(10)
```

```python
df = df.drop(columns=['name'])
df.head(10)
```

Comment: Name column is dropped beacause it is identifier feature

```python
df_encoded = pd.get_dummies(
    df,
    columns=['city'],
    dtype=int
)

df_encoded.head(10)
```

```python
target_col = "loan_approved" 

X =  df_encoded.drop(columns=[target_col])
Y = df_encoded[target_col]

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=.25,random_state=42)
```

```python
# Standard Scaling

scaler_std = StandardScaler()

X_train_std = scaler_std.fit_transform(X_train)
X_test_std = scaler_std.transform(X_test)

X_test_std_df = pd.DataFrame(X_test_std,columns=X_test.columns,index=X_test.index)
X_train_std_df = pd.DataFrame(X_train_std,columns=X_train.columns,index=X_train.index)
```

```python
display(X_train_std_df.head(10))
```

```python
display(X_test_std_df.head(10))
```

```python
# MinMax Scaling

scaler_mm = MinMaxScaler()

X_train_mm = scaler_mm.fit_transform(X_train)
X_test_mm = scaler_mm.transform(X_test)


X_train_mm_df = pd.DataFrame(X_train_mm,columns=X_train.columns,index=X_train.index)
X_test_mm_df = pd.DataFrame(X_test_mm,columns=X_test.columns,index=X_test.index)
```
